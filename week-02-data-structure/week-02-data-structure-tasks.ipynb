Task 1: Student Score Evaluator
Core Concepts: dict, for loop, if/elif/else, arithmetic operators.

Problem:
You have a classroom record where each student's name is mapped to their exam score (e.g., {"Alex": 88, "Jordan": 55, "Taylor": 72}).

Student Instructions:

Loop through each student in the dictionary.

Use conditionals to categorize each student:

Score 70 or above: Print "Pass"

Score below 70: Print "Fail"

Calculate and display the class average score.

Find and print the name of the student with the highest score.

```python
classroom_record = {
    "Alex" : 88, "John" : 55, "Stone" : 84, "Shaw" : 77, "Gabriel" : 91
}
highest_score = None
total_score = 0
highest_name = None
for name, score in classroom_record.items():
    if score >= 70:
        print(name,'your score is:',score, 'you passed')
    else:
        print(name,'your score is:',score,'you failed')
    total_score += score
if (highest_score is None or score > highest_score):
    highest_score = score
    highest_name = name
average = total_score / len(classroom_record)
print('class average is:',round(average,2))
print('top student is:',highest_name, 'with',highest_score)
```

Task 2: Number Categorizer & Filter
Core Concepts: list, for loop, modulo operator (%), if/elif/else.

Problem:
Given a raw list of integers (positive, negative, and zero), such as [12, -7, 0, 15, -2, 8, 0, 21].

Student Instructions:

Create three separate empty lists: evens, odds, and zeros.

Iterate through the number list:

If the number is 0, append it to zeros.

If the number is even (divisible by 2), append it to evens.

Otherwise, append it to odds.

Print the final counts and contents of each list.

```python
num = [12,-7,0,15,-2,8,0,21]
evens = []
odds = []
zeros = []
print('number is:',num)
for x in num:
    if(x == 0):
        zeros.append(x)
    elif(x % 2 == 0):
        evens.append(x)
    else:
        odds.append(x)
print('even numbers are:',evens,'and are:',len(evens))
print('odd numbers are:',odds,'and are:',len(odds))
print('zeros are:',zeros,'and are:',len(zeros))
```

Task 3: Interactive Number Guessing Game
Core Concepts: while loop, user input (input()), if/elif/else, break.

Problem:
Build a console game where the user tries to guess a predefined secret number (e.g., secret_number = 42).

Student Instructions:

Use a while True loop to prompt the user to guess the number.

Convert the input to an integer.

Provide feedback:

If the guess is higher than the secret number, print "Too high, try again!"

If the guess is lower, print "Too low, try again!"

If correct, congratulate the player, print how many attempts it took, and use break to stop the loop.


```python
secret_number = 42
attempt = 0
while True:
    x =int(input('Guess the secret code'))
    attempt += 1
    if(x == secret_number):
        print('Congratulations! you got it. it took you',attempt,'tries')
        break
    elif(x > secret_number):
        print('Too high, try again!')
    else:
        print('Too low, try again!')
```

Task 4: Duplicate Cleaner & Shared Hobby Finder
Core Concepts: set, list, set operations (intersection, union).

Problem:
Two friends have shared their hobby lists, but each list contains repeated entries and inconsistent typing (e.g., ["reading", "hiking", "reading", "gaming"] and ["cooking", "hiking", "cycling"]).

Student Instructions:

Convert both lists to sets to automatically remove all duplicates.

Find and print:

All unique hobbies between both friends combined (union).

Common hobbies that both friends enjoy (intersection).

Exclusive hobbies that the first friend enjoys but the second does not (difference).

```python
Ade = ['reading','hiking','gaming','reading','cooking','shopping','gaming']
Tope = ['running','gaming','reading','cooking','running','preaching','sleeping','preaching']
set1 = set(Ade)
set2 = set(Tope)
print("Ade's hobbies are:",set1)
print("Tope's hobbies are:",set2)
print(set1 | set2,"is all their hobbies combined.")
print(set1 & set2,"is common to both of them.")
print(set1 - set2,"are hobbies only Ade enjoys.")
```

```python
Task 5: Shopping Basket Bill Calculator
Core Concepts: list of tuples, dict, for loop, basic math.

Problem:
A mini-market has a price catalog dictionary: {"apple": 1.5, "bread": 2.0, "milk": 3.0}. 
A customer's basket is a list of tuples containing item names and quantities: [("apple", 3), ("milk", 2), ("snack", 1)].

Student Instructions:

Initialize a total_cost = 0.

Loop through each tuple in the basket.

Check if the item exists in the price catalog:

If it exists, multiply the unit price by the quantity and add it to total_cost.

If it does not exist, print a warning: "Item [name] is not available" and continue.

Print the final payable bill.
```

```python
prices = {"Apple":1.5,"Bread":2.0,"Milk":3.0}
customer_basket = [("Apple",3),("Milk",2),("Snack",1)]
total_cost = 0
for name, quantity in customer_basket:
    if (name in prices):
        item_cost = prices[name]*quantity
        total_cost += item_cost
    else:
        print(name,'is not available.')
        continue
print('Final pay bill is:',total_cost)        
        
```
