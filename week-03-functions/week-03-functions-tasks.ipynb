Write a function called celsius_to_fahrenheit(celsius) that converts a temperature from Celsius to Fahrenheit.

```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = celsius * 9/5 + 32
    return fahrenheit
```

```python
celsius_to_fahrenheit(100)
```

Write a function called validate_password(password) that checks whether a password contains at least 8 characters.
If it does, return "Strong enough".
Otherwise, return "Too short".

```python
def validate_password(password):
    if ( len(password) >= 8):
        return "Strong enough"
    else:
        return "Too short"
password = input("Enter password")
validate_password(password)
```

Write a function called calculate_total(prices) that accepts a list of prices and returns their total.


```python
def calculate_total(prices):
    total = 0
    for price in prices:
        total += price
    return total
prices = [float(x) for x in input().split(",")]
calculate_total(prices)
```

Create a function called student_result(name, scores) that:
Accepts a student's name and a list of scores.
Calculates the average score.
Assigns a grade:
70 and above: A
60–69: B
50–59: C
45–49: D
Below 45: F
Returns the student's name, average, and grade.

```python
def student_result(name,scores):
    
    average = sum(scores)/ len(scores)
    if (average >= 70):
        grade = "A"
    elif(average >= 60 and average <70):
        grade = "B"
    elif(average >= 50 and average <60):
        grade = "C"
    elif(average >= 45 and average <50):
        grade = "D"
    else:
        grade = "F"
    return name,average,grade
student_result("Peter",[67,93,81,89,20])    
```
