1. Create a Student Class
Create a class called Student with:
- Properties: name, age, and grade.
- A method called introduce() that displays the student’s details.
Task: Create two student objects and call introduce() on each.
Example output:
My name is Ada. I am 15 years old, and I am in Grade 10.

```python
class Student():
    def __init__(self,name,age,grade):
        self.name = name
        self.age = age
        self.grade = grade
    def introduce(self):
        print(f"My name is {self.name}. I am {self.age} years old. I am in grade {self.grade}")
```

```python
result = Student("Ayo",10,5)
```

```python
result.introduce()
```

```python
result2 = Student("Ade",7,9)
```

```python
result2.introduce()
```

2. Create a Rectangle Class
Create a class called Rectangle with:
- Properties: length and width.
- A method called calculateArea() that returns the length multiplied by the width.
Task: Create two rectangle objects with different dimensions and display their areas.
Example:
A rectangle with length 5 and width 3 should have an area of 15.

```python
class Rectangle():
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def calculate_area(self):
        print(f"A rectangle with length {self.length} and width {self.width} should have an area of {self.length*self.width}. ")
```

```python
result = Rectangle(5,3)
```

```python
result.calculate_area()
```

```python
result2 = Rectangle(7,9)
```

```python
result2.calculate_area()
```

3. Create a Car Class
Create a class called Car with:
- Properties: brand, color, and speed. Start the speed at 0.
- A method called accelerate() that increases the speed by 10.
- A method called showDetails() that displays the car’s brand, color, and current speed.
Task: Create two car objects. Call accelerate() twice on the first car, then display the details of both cars.
Expected result:
The first car’s speed should be 20, and the second car’s speed should still be 0.

```python
class Car():
    def __init__(self,brand,color):
        self.brand = brand
        self.color = color
        self.speed = 0
    def accelerate(self):
        self.speed += 10
    def show_details(self):
        print(f"The car's brand is {self.brand} and its color is {self.color} and its speed is {self.speed}")
```

```python
car1 = Car("Toyota","Blue")
car1.accelerate()
car1.accelerate()
car1.show_details()
```

```python
car2 = Car("Honda","Black")
car2.show_details()
```
